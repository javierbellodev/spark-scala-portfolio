### Funciones y Currying

```scala
def calcularDescuento(porcentaje: Double)(precio: Double): Double = {
    precio - (precio * porcentaje / 100)
}

val descuento10 = calcularDescuento(10) _
val precio = 100.0

println(s"Precio: $precio, con 10% descuento: ${descuento10(precio)}")
```

### Comprehesion

```scala
val numeros = List(1, 2, 3, 4, 5)

val cuadrados = for (n <- numeros) yield n * n
println(s"Cuadrados: $cuadrados")

println()

val pares = for {
    n <- numeros
    if n % 2 == 0
} yield n

println(s"Pares: $pares")
```
