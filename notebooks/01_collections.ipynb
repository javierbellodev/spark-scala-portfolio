### Uso de Set 

```scala
val numeros = Set(1, 2, 3, 4, 4, 5)
println(s"Set original (sin duplicados): $numeros")

println(s"¿Contiene 3? ${numeros.contains(3)}") 

val nuevosNumeros = numeros + 6
println(s"Nuevo set con 6 agregado: $nuevosNumeros")
println(s"Set original permanece igual: $numeros")

println()

//operaciones conjuntos
val set1 = Set(1, 2, 3, 4)
val set2 = Set(3, 4, 5, 6)

println(s"Unión: ${set1 union set2}")
println(s"Intersección: ${set1 intersect set2}")
println(s"Diferencia (set1 - set2): ${set1 diff set2}")

println()

//recorrer
val frutas = Set("Manzana", "Banana", "Mango")

for (fruta <- frutas) {
    println(s"Fruta: $fruta")
}

println()

//transforamciones
val numerosTransform = Set(1, 2, 3, 4)

val cuadrados = numerosTransform.map(n => n * n)
println(s"Cuadrados: $cuadrados")
```

### Uso de Mapas (Diccionarios)

```scala
val edades = Map(
    "Ana" -> 25,
    "Luis" -> 30,
    "Pedro" -> 28
  )

println(s"Mapa original: $edades")

println(s"Edad de Ana: ${edades("Ana")}")
println(s"¿Existe Juan? ${edades.contains("Juan")}")

println(s"Edad de Juan (seguro): ${edades.getOrElse("Juan", "No encontrado")}")

println()

//agregar
val nuevasEdades = edades + ("Carlos" -> 35)
println(s"Nuevo mapa con Carlos: $nuevasEdades")

println()

//recorrer
for ((clave, valor) <- edades) {
    println(s"$clave tiene $valor años")
}

println()

//claves y valores
val claves = edades.keys
val valores = edades.values

println(s"Claves: $claves")
println(s"Valores: $valores")

//transformaciones
val edadesIncrementadas = edades.map {
  case (nombre, edad) => (nombre, edad + 1)
}

println()
//filtro
val mayores27 = edades.filter {
    case (_, edad) => edad > 27
}
 
println(s"Mayores de 27: $mayores27")
```

### Uso de Lista

```scala
val numeros = List(1, 2, 3, 4, 5)

println(s"Primer elemento (head): ${numeros.head}")
println(s"Resto de la lista (tail): ${numeros.tail}")
println(s"Último elemento: ${numeros.last}")
println(s"Elemento en índice 2: ${numeros(2)}")

println()

//agregar al inicio
val nuevaLista = 0 :: numeros
println(s"Agregar al inicio 0: $nuevaLista")

//agregar al final
val agregarFinal = numeros :+ 6
println(s"Agregar al final (:+): $agregarFinal")
println()

//concatenar de listas
val otraLista = List(6, 7, 8)
val listaUnida = numeros ::: otraLista
println(s"Concatenación : $listaUnida")

println()

//recorrer
for (n <- numeros) {
    println(s"Número: $n")
}

println()

//transformaciones
val cuadrados = numeros.map(n => n * n)
println(s"Cuadrados: $cuadrados")

val pares = numeros.filter(_ % 2 == 0)
println(s"Pares: $pares")

val sumaTotal = numeros.sum
println(s"Suma total: $sumaTotal")

println()

//utilidades
println(s"Tamaño de la lista: ${numeros.length}")
println(s"Lista invertida: ${numeros.reverse}")
println(s"¿Contiene 3?: ${numeros.contains(3)}")

val ordenDesc = numeros.sortWith(_ > _)
println(s"Orden descendente: $ordenDesc")
```

### Uso de Array

```scala
val numeros = Array(1, 2, 3, 4, 5)
println(s"Array original: ${numeros.mkString(", ")}")
println(s"Elemento en índice 2: ${numeros(2)}")

numeros(2) = 99

println()

for (n <- numeros) {
    println(s"Número: $n")
}

//transformaciones
val cuadrados = numeros.map(n => n * n)
println(s"Cuadrados: ${cuadrados.mkString(", ")}")

val pares = numeros.filter(_ % 2 == 0)
println(s"Pares: ${pares.mkString(", ")}")

val sumaTotal = numeros.sum
println(s"Suma total: $sumaTotal")

//operaciones
println(s"Tamaño del array: ${numeros.length}")
println(s"Array invertido: ${numeros.reverse.mkString(", ")}")
println(s"¿Contiene 99?: ${numeros.contains(99)}")

val ordenadoDesc = numeros.sorted(Ordering[Int].reverse)
println(s"Orden descendente: ${ordenadoDesc.mkString(", ")}")
```

### Uso de Vector

```scala
val numeros = Vector(1, 2, 3, 4, 5)
println(s"Vector original: $numeros")

println()

val agregarInicio = 0 +: numeros
println(s"Agregar al inicio (+:): $agregarInicio")

val agregarFinal = numeros :+ 6
println(s"Agregar al final (:+): $agregarFinal")

println()

val cuadrados = numeros.map(_ * 2)
println(s"Multiplicados por 2: $cuadrados")

val pares = numeros.filter(_ % 2 == 0)
println(s"Pares: $pares")

val suma = numeros.fold(0)(_ + _)
println(s"Suma total con fold: $suma")

println()

val actualizado = numeros.updated(2, 99)
println(s"Vector actualizado (índice 2 -> 99): $actualizado")
println(s"Vector original permanece igual: $numeros")

println()

case class Producto(nombre: String, precio: Double)

val productos = Vector(
    Producto("Laptop", 1000),
    Producto("Mouse", 50),
    Producto("Monitor", 300)
)

val totalInventario = productos
    .map(_.precio)
    .fold(0.0)(_ + _)

println(s"Total inventario: $totalInventario")
```
