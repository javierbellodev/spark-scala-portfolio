###  Estructuras

```scala
 trait Movible {
    def mover(distancia: Int): Unit
  }
 
  class Vehiculo(val nombre: String) extends Movible {
    private var posicion: Int = 0

    override def mover(distancia: Int): Unit = {
      posicion += distancia
      println(s"$nombre se movió $distancia metros.")
    }

    def obtenerPosicion: Int = posicion
  }
 
  val carro = new Vehiculo("Carro")
  
  println(s"Posición inicial: ${carro.obtenerPosicion}")
  carro.mover(10)
  carro.mover(5)
  carro.mover(15)
  carro.mover(4)

  println(s"Posición final: ${carro.obtenerPosicion}")
```
