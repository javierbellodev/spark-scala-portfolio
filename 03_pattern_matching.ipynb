### Uso de Option

```scala
val valorPresente: Option[Int] = Some(10)
val valorAusente: Option[Int] = None

println(s"Valor presente: $valorPresente")
println(s"Valor ausente: $valorAusente")


println(s"Valor con getOrElse (presente): ${valorPresente.getOrElse(0)}")
println(s"Valor con getOrElse (ausente): ${valorAusente.getOrElse(0)}")

 
valorPresente match {
    case Some(v) => println(s"El valor es $v")
    case None    => println("No hay valor")
}

//transformaciones

val duplicado = valorPresente.map(_ * 2)
println(s"Valor duplicado: $duplicado")

val duplicadoNone = valorAusente.map(_ * 2)
println(s"Duplicado cuando es None: $duplicadoNone")
```

### Estructura Match

```scala
def describir(valor: Any): String = valor match {
    case i: Int     => s"Es un entero: $i"
    case s: String  => s"Es un String: $s"
    case b: Boolean => s"Es un Boolean: $b"
    case _          => "Tipo desconocido"
  }

println(describir(10))
println(describir("Hola"))
println(describir(true))

println()

///Option y match
val valor: Option[Int] = Some(50)

valor match {
    case Some(v) => println(s"El valor es $v")
    case None    => println("No hay valor")
}

println()

//match con condiciones
val nota = 85

val resultado = nota match {
    case n if n >= 90 => "A"
    case n if n >= 80 => "B"
    case n if n >= 70 => "C"
    case _            => "F"
}

println(s"Calificación: $resultado")
```
